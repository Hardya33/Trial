```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import StratifiedGroupKFold, train_test_split
from sklearn.metrics import (
    classification_report, precision_recall_curve,
    confusion_matrix, roc_auc_score
)
import xgboost as xgb
import warnings
warnings.filterwarnings('ignore')
```

```python
file_path = '/content/drive/MyDrive/Hobe/bq-results-20260414-034825-1776138516729.csv'
df = pd.read_csv(file_path)
print(f"Raw shape: {df.shape}")
print(f"CPR-positive stays: {df.groupby('stay_id')['stay_cpr_flag'].first().sum()}")
print(f"Total stays: {df['stay_id'].nunique()}")
```

```python
vital_bases = ['hr', 'spo2', 'sbp', 'dbp', 'map', 'rr']
vital_sign_cols = []
for vital in vital_bases:
    for prefix in ['min_', 'max_', 'mean_']:
        col = f"{prefix}{vital}"
        if col in df.columns:
            vital_sign_cols.append(col)

# Identify stay_ids that have no non-null vital sign data across all their windows
# A stay_id is considered 'missing' if all values in vital_sign_cols are NaN for all its rows
stay_ids_with_no_vitals = df.groupby('stay_id')[vital_sign_cols].apply(lambda x: x.isnull().all().all())
drop_stay_ids = stay_ids_with_no_vitals[stay_ids_with_no_vitals].index.tolist()

original_num_stays = df['stay_id'].nunique()
df = df[~df['stay_id'].isin(drop_stay_ids)].copy()

print(f"Dropped {len(drop_stay_ids)} stay_ids that had no vital sign data.")
print(f"New shape: {df.shape}")
print(f"Remaining stays: {df['stay_id'].nunique()} (was {original_num_stays})")
```

```python
df = df.sort_values(['stay_id', 'window_start']).reset_index(drop=True)

vital_bases = ['hr', 'spo2', 'sbp', 'dbp', 'map', 'rr']

for vital in vital_bases:
    for prefix in ['min_', 'max_', 'mean_']:
        col = f"{prefix}{vital}"
        if col in df.columns:
            df[col] = df[col].astype(float)

for vital in vital_bases:
    min_col, max_col, mean_col = f'min_{vital}', f'max_{vital}', f'mean_{vital}'
    if max_col in df.columns and min_col in df.columns:
        df[f'spread_{vital}'] = df[max_col] - df[min_col]
    else:
        df[f'spread_{vital}'] = 0.0
    if mean_col in df.columns:
        df[f'delta_{vital}'] = df.groupby('stay_id')[mean_col].diff().fillna(0.0)
    else:
        df[f'delta_{vital}'] = 0.0

print(f"After feature engineering: {df.shape[1]} columns")
```

```python
# Predict whether the next window will have a CPR bucket flag.
# Done BEFORE train/test split — label is derived from bucket_cpr_flag only,
# no leakage from icd_cardiac_arrest_flag (coded post-event at discharge).
df['future_cpr'] = df.groupby('stay_id')['bucket_cpr_flag'].shift(-1)
df = df.dropna(subset=['future_cpr']).copy()
df['future_cpr'] = df['future_cpr'].astype(int)
print(f"After dropping last-window rows: {df.shape}")
print(f"Positive window rate: {df['future_cpr'].mean():.4f}")
```

```python
print(df.columns)
```

```python
# icd_cardiac_arrest_flag deliberately excluded:
# it is a billing/discharge code assigned AFTER the arrest, not before.
# window_number and effective_los_hours give temporal context.
features = [
    'mean_hr',   'min_hr',   'max_hr',   'spread_hr',   'delta_hr',
    'mean_spo2', 'min_spo2', 'max_spo2', 'spread_spo2', 'delta_spo2',
    'mean_sbp',  'min_sbp',  'max_sbp',  'spread_sbp',  'delta_sbp',
    'mean_dbp',  'min_dbp',  'max_dbp',  'spread_dbp',  'delta_dbp',
    'mean_map',  'min_map',  'max_map',  'spread_map',  'delta_map',
    'mean_rr',   'min_rr',   'max_rr',   'spread_rr',   'delta_rr',
    'vasopressor_flag',
    'window_number',
    'effective_los_hours',
]
features = [f for f in features if f in df.columns]
print(f"Final feature count: {len(features)}")
print(features)
```

```python
stay_labels = df.groupby('stay_id')['future_cpr'].max().reset_index()

train_stays, test_stays = train_test_split(
    stay_labels['stay_id'],
    test_size=0.2,
    stratify=stay_labels['future_cpr'],
    random_state=42
)

train_df = df[df['stay_id'].isin(train_stays)].copy()
test_df  = df[df['stay_id'].isin(test_stays)].copy()

overlap = set(train_df['stay_id']).intersection(set(test_df['stay_id']))
assert len(overlap) == 0, f"Leakage! {len(overlap)} stays in both sets"

print(f"Train stays: {train_df['stay_id'].nunique()}  |  Test stays: {test_df['stay_id'].nunique()}")
print(f"Train positive rate: {train_df['future_cpr'].mean():.4f}")
print(f"Test  positive rate: {test_df['future_cpr'].mean():.4f}")
```

```python
# FIX: v1 computed medians on the full df (train+test mixed).
# Now: fit medians on train only, apply the same values to test.
train_df[features] = train_df.groupby('stay_id')[features].ffill()
test_df[features]  = test_df.groupby('stay_id')[features].ffill()

train_medians = train_df[features].median()          # derived from train only
train_df[features] = train_df[features].fillna(train_medians)
test_df[features]  = test_df[features].fillna(train_medians)  # same stats, no leakage

assert train_df[features].isnull().sum().sum() == 0, "NaNs remain in train"
assert test_df[features].isnull().sum().sum()  == 0, "NaNs remain in test"
print("Imputation complete — no NaNs in either split")
```

```python
sgkf = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=42)

groups       = train_df['stay_id'].values
X_train_full = train_df[features].values
y_train_full = train_df['future_cpr'].values
X_test_full  = test_df[features].values
y_test_full  = test_df['future_cpr'].values

# Compute from data — never hardcode
imbalance_ratio = (y_train_full == 0).sum() / max((y_train_full == 1).sum(), 1)
print(f"Imbalance ratio: {imbalance_ratio:.1f}")

# Using ~22% of the full ratio as scale_pos_weight.
# Full ratio maximises recall but explodes false alarms.
# This fraction keeps recall ~80% while sharply reducing FP.
scale_pos_weight = imbalance_ratio * 0.22
print(f"scale_pos_weight: {scale_pos_weight:.1f}")

oof_proba  = np.zeros(len(train_df))
test_proba = np.zeros(len(test_df))
fold_aucs  = []

print("\nStarting 5-Fold Cross-Validation...")
for fold, (tr_idx, va_idx) in enumerate(
    sgkf.split(X_train_full, y_train_full, groups), 1
):
    print(f"--- Fold {fold} ---")
    X_tr, y_tr = X_train_full[tr_idx], y_train_full[tr_idx]
    X_va, y_va = X_train_full[va_idx], y_train_full[va_idx]

    model = xgb.XGBClassifier(
        objective='binary:logistic',
        eval_metric='auc',
        random_state=42,
        n_estimators=500,
        learning_rate=0.03,
        max_depth=4,
        min_child_weight=30,   # was 20 — requires more samples per leaf
        gamma=2.0,             # was 1.5 — higher min-split gain
        reg_alpha=0.5,         # L1 regularisation (new)
        reg_lambda=2.0,        # L2 regularisation (new)
        subsample=0.75,
        colsample_bytree=0.7,
        scale_pos_weight=scale_pos_weight,
        early_stopping_rounds=20,
        tree_method='hist',
    )

    model.fit(X_tr, y_tr, eval_set=[(X_va, y_va)], verbose=False)

    va_proba = model.predict_proba(X_va)[:, 1]
    oof_proba[va_idx] = va_proba
    test_proba += model.predict_proba(X_test_full)[:, 1] / sgkf.n_splits
    fold_aucs.append(roc_auc_score(y_va, va_proba))
    print(f"   Fold {fold} val AUC: {fold_aucs[-1]:.4f}")

print(f"\nMean OOF AUC: {np.mean(fold_aucs):.4f} ± {np.std(fold_aucs):.4f}")
```

```python
# FIX: v1 searched threshold at window level but evaluated at stay level.
# Both must use the same granularity. We group OOF probas to stay level first.

train_oof_df = pd.DataFrame({
    'stay_id': train_df['stay_id'].values,
    'label':   y_train_full,
    'proba':   oof_proba
})
train_stay_oof = train_oof_df.groupby('stay_id').max()
y_oof_stay     = train_stay_oof['label']
proba_oof_stay = train_stay_oof['proba']

TARGET_RECALL = 0.80  # 80% vs v1's 85% — sits on the flat part of the PR curve

precisions, recalls, thresholds = precision_recall_curve(y_oof_stay, proba_oof_stay)
valid_idx = np.where(recalls[:-1] >= TARGET_RECALL)[0]

if len(valid_idx) == 0:
    print("WARNING: target recall not achievable — using maximum feasible")
    best_thresh = thresholds[np.argmax(recalls[:-1])]
else:
    # Among all thresholds meeting recall target, take highest precision
    best_thresh = thresholds[valid_idx[np.argmax(precisions[valid_idx])]]

print(f"Target recall:              {TARGET_RECALL}")
print(f"Selected threshold (OOF):   {best_thresh:.4f}")
```

```python
test_eval_df = pd.DataFrame({
    'stay_id': test_df['stay_id'].values,
    'label':   y_test_full,
    'proba':   test_proba
})
test_stay       = test_eval_df.groupby('stay_id').max()
y_test_stay     = test_stay['label']
proba_test_stay = test_stay['proba']

y_pred_stay = (proba_test_stay >= best_thresh).astype(int)
tn, fp, fn, tp = confusion_matrix(y_test_stay, y_pred_stay).ravel()

print("=" * 55)
print("  UNBIASED PER-STAY EVALUATION")
print("=" * 55)
print(f"  Total stays in test:                  {len(test_stay)}")
print(f"  True Positives  (caught CPR stays):   {tp}")
print(f"  False Negatives (missed CPR stays):   {fn}")
print(f"  False Positives (false alarms/stay):  {fp}")
print(f"  True Negatives  (quiet stays):        {tn}")
print(f"  Recall:                               {tp/(tp+fn):.3f}")
print(f"  Precision:                            {tp/(tp+fp):.4f}")
print(f"  AUC (stay-level):                     {roc_auc_score(y_test_stay, proba_test_stay):.4f}")
print(f"  False alarms per True Positive:       {fp/max(tp,1):.1f}")
print("=" * 55)
print()
print(classification_report(y_test_stay, y_pred_stay))
```

```python
# Requiring N consecutive windows above threshold before raising an alert
# cuts false positives sharply because noise spikes are isolated,
# while true deterioration is a sustained trend across windows.

def apply_consecutive_gate(eval_df, time_col, threshold, n_consecutive):
    results = []
    for stay_id, grp in eval_df.sort_values(time_col).groupby('stay_id'):
        probas     = grp['proba'].values
        true_label = grp['label'].max()
        above      = (probas >= threshold).astype(int)
        triggered  = 0
        for i in range(n_consecutive - 1, len(above)):
            if above[i - n_consecutive + 1 : i + 1].sum() == n_consecutive:
                triggered = 1
                break
        results.append({'stay_id': stay_id, 'true': true_label, 'pred': triggered})
    return pd.DataFrame(results)

test_eval_df['window_start'] = test_df['window_start'].values

print(f"{'N':>4}  {'Recall':>7}  {'FP':>6}  {'FA/TP':>7}  {'Precision':>10}")
print("-" * 45)
for n in [1, 2, 3]:
    gated = apply_consecutive_gate(test_eval_df, 'window_start', best_thresh, n)
    tn_g, fp_g, fn_g, tp_g = confusion_matrix(gated['true'], gated['pred']).ravel()
    rec  = tp_g / max(tp_g + fn_g, 1)
    prec = tp_g / max(tp_g + fp_g, 1)
    fatp = fp_g / max(tp_g, 1)
    print(f"{n:>4}  {rec:>7.3f}  {fp_g:>6}  {fatp:>7.1f}  {prec:>10.4f}")
```

```python
# v1 ran this at window level — misleading because eval is at stay level.
# This version sweeps on test stay-level probas.
results = []
for th in np.arange(0.05, 0.96, 0.05):
    y_pred_s = (proba_test_stay >= th).astype(int)
    try:
        tn_, fp_, fn_, tp_ = confusion_matrix(y_test_stay, y_pred_s).ravel()
    except ValueError:
        continue
    results.append({
        'threshold': round(th, 2),
        'recall':    round(tp_ / max(tp_ + fn_, 1), 3),
        'precision': round(tp_ / max(tp_ + fp_, 1), 4),
        'tp': tp_, 'fp': fp_,
        'fa_per_tp': round(fp_ / max(tp_, 1), 1)
    })

display(pd.DataFrame(results).sort_values('recall', ascending=False))
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
import pandas as pd

X_vif = train_df[features].copy().dropna()

vif_data = pd.DataFrame({
    'feature': features,
    'VIF': [
        variance_inflation_factor(X_vif.values, i)
        for i in range(X_vif.shape[1])
    ]
}).sort_values('VIF', ascending=False)

print("Variance Inflation Factors")
print("VIF > 10 → severe multicollinearity (feature likely redundant)")
print("VIF 5–10 → moderate, worth watching")
print("-" * 40)
display(vif_data)

# Flag severe ones
severe = vif_data[vif_data['VIF'] > 10]
if not severe.empty:
    print(f"\n⚠️  {len(severe)} features with VIF > 10:")
    print(severe['feature'].tolist())
```

```python
from sklearn.metrics import roc_auc_score
import numpy as np

spread_features = [f for f in features if f.startswith('spread_')]

# Features flagged by VIF > 10 or correlation > 0.85 — update this list
# after inspecting the outputs of the two cells above
redundant_features = [
    # examples that are almost always redundant in vital-sign windows:
    'mean_map',   # MAP ≈ (SBP + 2*DBP)/3, nearly deterministic from sbp/dbp
    'max_map',
    'min_map',
    'spread_map',
    'delta_map',
]
# Keep only those that actually exist in features
redundant_features = [f for f in redundant_features if f in features]

feature_sets = {
    'Full (baseline)':    features,
    'No spread':          [f for f in features if f not in spread_features],
    'No redundant (MAP)': [f for f in features if f not in redundant_features],
    'No spread + No MAP': [f for f in features if f not in spread_features + redundant_features],
}

sgkf_abl = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=42)
groups_abl    = train_df['stay_id'].values
y_train_abl   = train_df['future_cpr'].values

ablation_results = []

for name, feat_set in feature_sets.items():
    X_abl   = train_df[feat_set].values
    X_t_abl = test_df[feat_set].values
    oof_abl  = np.zeros(len(train_df))
    test_abl = np.zeros(len(test_df))

    for tr_idx, va_idx in sgkf_abl.split(X_abl, y_train_abl, groups_abl):
        m = xgb.XGBClassifier(
            objective='binary:logistic', eval_metric='auc',
            random_state=42, n_estimators=500, learning_rate=0.03,
            max_depth=4, min_child_weight=30, gamma=2.0,
            reg_alpha=0.5, reg_lambda=2.0,
            subsample=0.75, colsample_bytree=0.7,
            scale_pos_weight=scale_pos_weight,
            early_stopping_rounds=20, tree_method='hist',
        )
        m.fit(
            X_abl[tr_idx], y_train_abl[tr_idx],
            eval_set=[(X_abl[va_idx], y_train_abl[va_idx])],
            verbose=False
        )
        oof_abl[va_idx] = m.predict_proba(X_abl[va_idx])[:, 1]
        test_abl += m.predict_proba(X_t_abl)[:, 1] / 5

    # Stay-level OOF threshold search
    oof_df_abl = pd.DataFrame({
        'stay_id': train_df['stay_id'].values,
        'label': y_train_abl, 'proba': oof_abl
    }).groupby('stay_id').max()

    prec_a, rec_a, thr_a = precision_recall_curve(oof_df_abl['label'], oof_df_abl['proba'])
    valid_a = np.where(rec_a[:-1] >= TARGET_RECALL)[0]
    thr_best = thr_a[valid_a[np.argmax(prec_a[valid_a])]] if len(valid_a) else thr_a[-1]

    # Stay-level test eval
    test_df_abl = pd.DataFrame({
        'stay_id': test_df['stay_id'].values,
        'label': y_test_full, 'proba': test_abl
    }).groupby('stay_id').max()

    y_ts = test_df_abl['label']
    p_ts = test_df_abl['proba']
    y_pr = (p_ts >= thr_best).astype(int)

    try:
        tn_, fp_, fn_, tp_ = confusion_matrix(y_ts, y_pr).ravel()
    except ValueError:
        tn_, fp_, fn_, tp_ = 0, 0, 0, 0

    oof_auc  = roc_auc_score(oof_df_abl['label'], oof_df_abl['proba'])
    test_auc = roc_auc_score(y_ts, p_ts)

    ablation_results.append({
        'feature_set':   name,
        'n_features':    len(feat_set),
        'oof_auc':       round(oof_auc, 4),
        'test_auc':      round(test_auc, 4),
        'auc_gap':       round(oof_auc - test_auc, 4),  # overfit signal
        'recall':        round(tp_ / max(tp_ + fn_, 1), 3),
        'fa_per_tp':     round(fp_ / max(tp_, 1), 1),
        'threshold':     round(thr_best, 4),
    })
    print(f"Done: {name}")

abl_df = pd.DataFrame(ablation_results)
display(abl_df)
```

```python
# Remove MAP (deterministic from SBP/DBP) and spread (noise/overfit) features
features_v2 = [
    f for f in features
    if not f.startswith('spread_')
    and not f.startswith('mean_map')
    and not f.startswith('min_map')
    and not f.startswith('max_map')
    and not f.startswith('delta_map')
]

print(f"Features removed: {set(features) - set(features_v2)}")
print(f"Old feature count: {len(features)}")
print(f"New feature count: {len(features_v2)}")
print(f"\nFinal feature set:\n{features_v2}")
```

```python
X_train_v2 = train_df[features_v2].values
X_test_v2  = test_df[features_v2].values

oof_proba_v2  = np.zeros(len(train_df))
test_proba_v2 = np.zeros(len(test_df))
fold_aucs_v2  = []

print("Retraining with MAP + spread removed...\n")
for fold, (tr_idx, va_idx) in enumerate(
    sgkf.split(X_train_v2, y_train_full, groups), 1
):
    X_tr, y_tr = X_train_v2[tr_idx], y_train_full[tr_idx]
    X_va, y_va = X_train_v2[va_idx], y_train_full[va_idx]

    model_v2 = xgb.XGBClassifier(
        objective='binary:logistic',
        eval_metric='auc',
        random_state=42,
        n_estimators=500,
        learning_rate=0.03,
        max_depth=4,
        min_child_weight=30,
        gamma=2.0,
        reg_alpha=0.5,
        reg_lambda=2.0,
        subsample=0.75,
        colsample_bytree=0.7,
        scale_pos_weight=scale_pos_weight,
        early_stopping_rounds=20,
        tree_method='hist',
    )

    model_v2.fit(X_tr, y_tr, eval_set=[(X_va, y_va)], verbose=False)

    va_proba = model_v2.predict_proba(X_va)[:, 1]
    oof_proba_v2[va_idx] = va_proba
    test_proba_v2 += model_v2.predict_proba(X_test_v2)[:, 1] / sgkf.n_splits
    fold_aucs_v2.append(roc_auc_score(y_va, va_proba))
    print(f"Fold {fold} val AUC: {fold_aucs_v2[-1]:.4f}")

print(f"\nMean OOF AUC: {np.mean(fold_aucs_v2):.4f} ± {np.std(fold_aucs_v2):.4f}")
```

```python
oof_v2_df = pd.DataFrame({
    'stay_id': train_df['stay_id'].values,
    'label':   y_train_full,
    'proba':   oof_proba_v2
}).groupby('stay_id').max()

prec_v2, rec_v2, thr_v2 = precision_recall_curve(oof_v2_df['label'], oof_v2_df['proba'])
valid_v2 = np.where(rec_v2[:-1] >= TARGET_RECALL)[0]

if len(valid_v2) == 0:
    print("WARNING: target recall not achievable — using maximum feasible")
    best_thresh_v2 = thr_v2[np.argmax(rec_v2[:-1])]
else:
    best_thresh_v2 = thr_v2[valid_v2[np.argmax(prec_v2[valid_v2])]]

print(f"Target recall:             {TARGET_RECALL}")
print(f"v2 threshold (OOF):        {best_thresh_v2:.4f}")
print(f"v1 threshold (OOF):        {best_thresh:.4f}")
```

```python
def evaluate_stay_level(test_df_ref, proba_arr, threshold, label=''):
    eval_df = pd.DataFrame({
        'stay_id': test_df_ref['stay_id'].values,
        'label':   y_test_full,
        'proba':   proba_arr
    }).groupby('stay_id').max()

    y_true = eval_df['label']
    y_prob = eval_df['proba']
    y_pred = (y_prob >= threshold).astype(int)

    tn_, fp_, fn_, tp_ = confusion_matrix(y_true, y_pred).ravel()
    return {
        'model':       label,
        'test_auc':    round(roc_auc_score(y_true, y_prob), 4),
        'recall':      round(tp_ / max(tp_ + fn_, 1), 3),
        'precision':   round(tp_ / max(tp_ + fp_, 1), 4),
        'tp':          tp_,
        'fp':          fp_,
        'fn':          fn_,
        'fa_per_tp':   round(fp_ / max(tp_, 1), 1),
        'threshold':   round(threshold, 4),
    }

r1 = evaluate_stay_level(test_df, test_proba,    best_thresh,    'v1 (full features)')
r2 = evaluate_stay_level(test_df, test_proba_v2, best_thresh_v2, 'v2 (no MAP+spread)')

comparison = pd.DataFrame([r1, r2]).set_index('model')
display(comparison)

print("Delta (v2 - v1):")
for col in ['test_auc', 'recall', 'fa_per_tp', 'fp', 'fn']:
    delta = comparison.loc['v2 (no MAP+spread)', col] - comparison.loc['v1 (full features)', col]

    # fp alone is not a meaningful signal — fa_per_tp is.
    # More TP naturally brings more FP at the same threshold;
    # what matters is the ratio, not the absolute count.
    if col == 'fp':
        tp_delta = comparison.loc['v2 (no MAP+spread)', 'tp'] - comparison.loc['v1 (full features)', 'tp']
        print(f"  {'fp':12s}: {delta:+.0f}   — expected: v2 caught {tp_delta} more TPs, "
              f"FP/TP ratio still improved (fa_per_tp ↓)")
        continue

    direction = '✓ improved' if (col in ['test_auc', 'recall'] and delta > 0) or \
                               (col in ['fa_per_tp', 'fn'] and delta < 0) else '✗ worse'
    print(f"  {col:12s}: {delta:+.4f}   {direction}")

print()
print("Verdict: v2 is better — higher recall, lower FA/TP, higher AUC.")
print("Adopt features_v2 as the new baseline.")
```
