```python
!pip install tensorflow

import os
import random
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns # Added seaborn import
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    roc_auc_score, confusion_matrix,
    classification_report, PrecisionRecallDisplay
)
from sklearn.calibration import CalibrationDisplay
from tensorflow import keras
from tensorflow.keras import layers
import warnings
warnings.filterwarnings('ignore')

SEED = 42
os.environ['PYTHONHASHSEED'] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# Enable XLA JIT compilation globally to bypass CuDNN penalties
tf.config.optimizer.set_jit(True)

# ── Tunable knobs ─────────────────────────────────────
SEQUENCE_LEN   = 24       # 24 × 30min = 12 hours
BATCH_SIZE     = 64
EPOCHS         = 60
LR             = 1e-3
HIDDEN         = 64
DROPOUT        = 0.3
MAX_FA_PER_TP  = 7.0      # Clinical constraint for alarm fatigue
# ──────────────────────────────────────────────────────

print(f"TF version : {tf.__version__}")
print(f"GPU        : {tf.config.list_physical_devices('GPU')}")
```

```python
file_path = '/content/drive/MyDrive/bq-results-20260408-160350-1775666955037/bq-results-20260408-160350-1775666955037.csv'
df = pd.read_csv(file_path, low_memory=False)
df['window_start'] = pd.to_datetime(df['window_start'])
df['window_end']   = pd.to_datetime(df['window_end'])

vital_bases = ['hr', 'spo2', 'sbp', 'dbp', 'rr']

# Safely convert types
for vital in vital_bases:
    for prefix in ['min_', 'max_', 'mean_']:
        col = f"{prefix}{vital}"
        if col in df.columns:
            df[col] = df[col].astype(float)

# Calculate Deltas
for vital in vital_bases:
    mean_col = f'mean_{vital}'
    if mean_col in df.columns:
        df[f'delta_{vital}'] = df.groupby('stay_id')[mean_col].diff().fillna(0.0)
    else:
        df[f'delta_{vital}'] = 0.0

# EXTRACT MISSING INDICATORS BEFORE IMPUTATION
missing_features = []
for vital in vital_bases:
    indicator_col = f'{vital}_sensor_missing'
    df[indicator_col] = df[f'mean_{vital}'].isna().astype(float)
    missing_features.append(indicator_col)

# Compile Master Feature List
features = [
    'mean_hr',   'min_hr',   'max_hr',   'delta_hr',
    'mean_spo2', 'min_spo2', 'max_spo2', 'delta_spo2',
    'mean_sbp',  'min_sbp',  'max_sbp',  'delta_sbp',
    'mean_dbp',  'min_dbp',  'max_dbp',  'delta_dbp',
    'mean_rr',   'min_rr',   'max_rr',   'delta_rr',
    'vasopressor_flag', 'window_number', 'effective_los_hours'
]
features = [f for f in features if f in df.columns] + missing_features

print(f"Total features : {len(features)}")
```

```python
stay_labels = df.groupby('stay_id')['stay_cpr_flag'].first().reset_index()

# 15% Unseen Test Set
train_val_stays, test_stays = train_test_split(
    stay_labels['stay_id'], test_size=0.15, stratify=stay_labels['stay_cpr_flag'], random_state=SEED
)

# 70% Train, 15% Val
train_val_labels = stay_labels[stay_labels['stay_id'].isin(train_val_stays)]
train_stays, val_stays = train_test_split(
    train_val_stays, test_size=0.15 / 0.85, stratify=train_val_labels['stay_cpr_flag'], random_state=SEED
)

train_df = df[df['stay_id'].isin(train_stays)].copy()
val_df   = df[df['stay_id'].isin(val_stays)].copy()
test_df  = df[df['stay_id'].isin(test_stays)].copy()

# Forward Fill & Median Imputation (Strictly derived from Train)
train_df[features] = train_df.groupby('stay_id')[features].ffill()
val_df[features]   = val_df.groupby('stay_id')[features].ffill()
test_df[features]  = test_df.groupby('stay_id')[features].ffill()

train_medians      = train_df[features].median()
train_df[features] = train_df[features].fillna(train_medians)
val_df[features]   = val_df[features].fillna(train_medians)
test_df[features]  = test_df[features].fillna(train_medians)

# Scaling
scaler = StandardScaler()
train_df[features] = scaler.fit_transform(train_df[features])
val_df[features]   = scaler.transform(val_df[features])
test_df[features]  = scaler.transform(test_df[features])

def build_sequences(data_df, feature_cols, seq_len):
    X, y, ids = [], [], []
    for stay_id, grp in data_df.sort_values('window_start').groupby('stay_id'):
        seq = grp[feature_cols].values.astype(np.float32)
        label = int(grp['stay_cpr_flag'].max())
        if len(seq) >= seq_len:
            seq = seq[-seq_len:]
        else:
            pad = np.zeros((seq_len - len(seq), len(feature_cols)), dtype=np.float32)
            seq = np.vstack([pad, seq])
        X.append(seq); y.append(label); ids.append(stay_id)
    return np.array(X), np.array(y), np.array(ids)

X_train, y_train, train_ids = build_sequences(train_df, features, SEQUENCE_LEN)
X_val,   y_val,   val_ids   = build_sequences(val_df,   features, SEQUENCE_LEN)
X_test,  y_test,  test_ids  = build_sequences(test_df,  features, SEQUENCE_LEN)

print(f"X_train shape: {X_train.shape} | y_train positive rate: {y_train.mean():.4f}")
```

```python
class MaskedAttention(layers.Layer):
    def __init__(self, seed=SEED, **kwargs):
        super(MaskedAttention, self).__init__(**kwargs)
        self.seed = seed

    def build(self, input_shape):
        self.W = self.add_weight(shape=(input_shape[-1], 1), initializer=tf.keras.initializers.GlorotUniform(seed=self.seed), trainable=True, name='attn_W')
        self.b = self.add_weight(shape=(1,), initializer='zeros', trainable=True, name='attn_b')

    def call(self, inputs, mask=None):
        score = tf.tensordot(inputs, self.W, axes=1) + self.b
        score = tf.nn.tanh(score)
        if mask is not None:
            mask = tf.cast(mask, inputs.dtype)
            mask = tf.expand_dims(mask, axis=-1)
            score += (1.0 - mask) * -10000.0 # Prevents XLA overflow
        attention_weights = tf.nn.softmax(score, axis=1)
        context = tf.reduce_sum(attention_weights * inputs, axis=1)
        return context, attention_weights

    def compute_mask(self, inputs, mask=None):
        return None

def build_lstm(seq_len, n_features, hidden=64, dropout=0.3):
    inputs = keras.Input(shape=(seq_len, n_features), name='vital_sequence')
    x = layers.Masking(mask_value=0.0, name='masking')(inputs)
    x = layers.SpatialDropout1D(dropout, name='spatial_drop_1')(x)
    x = layers.LSTM(hidden, return_sequences=True, name='lstm_1', use_cudnn=False)(x) # Set use_cudnn=False
    x = layers.LayerNormalization(name='ln_1')(x)
    x = layers.SpatialDropout1D(dropout, name='spatial_drop_2')(x)
    x = layers.LSTM(hidden // 2, return_sequences=True, name='lstm_2', use_cudnn=False)(x) # Set use_cudnn=False
    x = layers.LayerNormalization(name='ln_2')(x)

    context, attn_weights = MaskedAttention(seed=SEED, name='masked_attention')(x)

    x = layers.Dense(64, activation='relu', name='fc_1')(context)
    x = layers.Dropout(dropout, seed=SEED)(x)
    out = layers.Dense(1, activation='sigmoid', dtype='float32', name='output')(x)

    model = keras.Model(inputs, out, name='LSTM_Model')
    attn_extractor = keras.Model(inputs, attn_weights, name='Attention_Extractor')
    return model, attn_extractor
model_lstm, attn_model = build_lstm(SEQUENCE_LEN, len(features), HIDDEN, DROPOUT)
model_lstm.summary()
```

```python
model_lstm.compile(
    optimizer = keras.optimizers.Adam(learning_rate=LR),
    loss      = keras.losses.BinaryFocalCrossentropy(gamma=2.0, alpha=0.60),
    metrics   = [keras.metrics.AUC(name='auc'), keras.metrics.Precision(name='precision'), keras.metrics.Recall(name='recall')]
)

train_dataset = tf.data.Dataset.from_tensor_slices((X_train, y_train)).shuffle(2048, seed=SEED).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
val_dataset   = tf.data.Dataset.from_tensor_slices((X_val, y_val)).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

callbacks = [
    keras.callbacks.EarlyStopping(monitor='val_auc', mode='max', patience=10, restore_best_weights=True),
    keras.callbacks.ReduceLROnPlateau(monitor='val_auc', mode='max', patience=5, factor=0.5, min_lr=1e-5)
]

history = model_lstm.fit(train_dataset, validation_data=val_dataset, epochs=EPOCHS, callbacks=callbacks, verbose=1)

# Plot Training History
fig, axes = plt.subplots(1, 3, figsize=(18, 4))
for ax, metric in zip(axes, ['loss', 'auc', 'recall']):
    ax.plot(history.history[metric], label='Train')
    ax.plot(history.history[f'val_{metric}'], label='Val')
    ax.set_title(metric.upper()); ax.set_xlabel('Epoch'); ax.legend(); ax.grid(True, alpha=0.3)
plt.suptitle('Training Curves (Focal Loss + Missing Indicators)', y=1.05)
plt.show()
```

```python
print("\n" + "=" * 60)
print(" 📈 TRAINING SET EVALUATION (Original Model)")
print("=" * 60)

# --- Code to calculate best_thresh (copied from cell vVSQJr_UKd8h) ---
# This section is added to ensure 'best_thresh' is defined before use in this cell.
# Ideally, this would be computed once on the validation set in an earlier cell.
val_proba = model_lstm.predict(X_val, verbose=0).flatten()
best_recall, best_thresh = 0.0, 0.5

for th in np.arange(0.01, 1.0, 0.01):
    y_v = (val_proba >= th).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_val, y_v, labels=[0, 1]).ravel()
    if tp == 0: continue
    if (fp / tp) <= MAX_FA_PER_TP and (tp / (tp + fn)) > best_recall:
        best_recall = tp / (tp + fn)
        best_thresh = th
# --- End of best_thresh calculation code ---

train_proba = model_lstm.predict(X_train, verbose=0).flatten()
y_train_pred = (train_proba >= best_thresh).astype(int) # Use the best_thresh derived from validation set

# Calculate metrics for the training set
tn_train, fp_train, fn_train, tp_train = confusion_matrix(y_train, y_train_pred, labels=[0, 1]).ravel()

train_recall = tp_train / max(tp_train + fn_train, 1)
train_fa_ratio = fp_train / max(tp_train, 1)
overall_train_auc = roc_auc_score(y_train, train_proba)

print(f"Operating Threshold : {best_thresh:.3f}")
print(f"Overall FA/TP Ratio : {train_fa_ratio:.1f} False Alarms per True Positive\n")
print(classification_report(y_train, y_train_pred, target_names=['Stable (0)', 'Arrest (1)']))
print(f"Overall AUC         : {overall_train_auc:.3f}")
print("=" * 60)
```

```python
# 1. Objective Function: Maximize Recall strictly bounded by FA/TP limit
val_proba = model_lstm.predict(X_val, verbose=0).flatten()
best_recall, best_thresh = 0.0, 0.5

for th in np.arange(0.01, 1.0, 0.01):
    y_v = (val_proba >= th).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_val, y_v, labels=[0, 1]).ravel()
    if tp == 0: continue
    if (fp / tp) <= MAX_FA_PER_TP and (tp / (tp + fn)) > best_recall:
        best_recall = tp / (tp + fn)
        best_thresh = th

print(f"Algorithmic Validation Threshold: {best_thresh:.4f}")

# 2. Final Test Set Evaluation
test_proba = model_lstm.predict(X_test, verbose=0).flatten()
y_test_pred = (test_proba >= best_thresh).astype(int)

tn, fp, fn, tp = confusion_matrix(y_test, y_test_pred, labels=[0, 1]).ravel()

print("\n" + "=" * 50)
print("  FINAL UNBIASED TEST EVALUATION")
print("=" * 50)
print(f"  Recall (Sensitivity) : {tp/max(tp+fn, 1):.3f}")
print(f"  FA per TP Ratio      : {fp/max(tp, 1):.1f}")
print(f"  Confusion Matrix     : TP:{tp}  FP:{fp}  FN:{fn}  TN:{tn}")
print("=" * 50)

# Unbiased Test PR Curve & Calibration Curve
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

PrecisionRecallDisplay.from_predictions(y_test, test_proba, ax=ax1, name="LSTM", color='steelblue')
ax1.scatter([tp/max(tp+fn, 1)], [tp/max(tp+fp, 1)], color='red', s=80, label=f'Operating Point ({best_thresh:.2f})')
ax1.set_title('Test Set Precision-Recall Curve'); ax1.legend(); ax1.grid(True, alpha=0.3)

CalibrationDisplay.from_predictions(y_test, test_proba, ax=ax2, n_bins=10, name="LSTM (Uncalibrated)")
ax2.set_title('Reliability Diagram (Calibration Check)')
ax2.grid(True, alpha=0.3)
plt.show()
```

```python
# ---------------------------------------------------------
# DEFENSIVE CHECKS & CALIBRATION NOTICES
# ---------------------------------------------------------
assert np.sum(y_test) > 0, "CRITICAL ERROR: No positive cases (CPR=1) in the Test Set!"

# NOTE ON CALIBRATION:
# The Focal Loss alpha parameter intentionally warps output probabilities to heavily
# penalize false negatives. Outputs are uncalibrated risk scores used strictly for
# rank-ordering and thresholding, not absolute probability estimation.
# Platt/Isotonic scaling is omitted as it does not alter rank-order metrics (AUC/PR).

# ---------------------------------------------------------
# ALGORITHMIC THRESHOLDING (Derived on Val, Applied to Test)
# ---------------------------------------------------------
# ... (Assuming best_thresh is derived here from val_dataset as before) ...
 # Example derived threshold

test_proba = model_lstm.predict(X_test, batch_size=BATCH_SIZE, verbose=0).flatten()
y_test_pred = (test_proba >= best_thresh).astype(int)

tn, fp, fn, tp = confusion_matrix(y_test, y_test_pred, labels=[0, 1]).ravel()
test_recall = tp / (tp + fn)
test_fa_ratio = fp / max(tp, 1)

# Soft Assertion for Operational Constraint (Allows 20% real-world variance)
if test_fa_ratio > (MAX_FA_PER_TP * 1.2):
    print(f"⚠️ WARNING: Test FA/TP ratio ({test_fa_ratio:.1f}) severely exceeded constraint ({MAX_FA_PER_TP})!")

print("\n" + "=" * 60)
print(" 🏥 FINAL CLINICAL EVALUATION (UNSEEN TEST SET)")
print("=" * 60)
print(f"Operating Threshold : {best_thresh:.3f}")
print(f"Overall FA/TP Ratio : {test_fa_ratio:.1f} False Alarms per True Positive\n")
print(classification_report(y_test, y_test_pred, target_names=['Stable (0)', 'Arrest (1)']))
print("=" * 60)

# ---------------------------------------------------------
# VECTORIZED FALSE NEGATIVE ANALYSIS (BATCHED INFERENCE)
# ---------------------------------------------------------
fn_indices = np.where((y_test == 1) & (y_test_pred == 0))[0]

if len(fn_indices) > 0:
    risk_90th = np.percentile(test_proba, 90)
    fn_records = []
    fn_seqs = X_test[fn_indices]
    fn_max_probs = []

    print(f"\nAnalyzing {len(fn_indices)} False Negatives via Batched Inference...")

    # BATCHED INFERENCE: Process all 24 windows for a sequence in a single GPU pass
    for seq in fn_seqs:
        incremental_windows = []
        for w in range(1, SEQUENCE_LEN + 1):
            # Pad the historical view to maintain fixed sequence length
            padded_view = np.vstack([np.zeros((SEQUENCE_LEN - w, seq.shape[1])), seq[SEQUENCE_LEN - w:]])
            incremental_windows.append(padded_view)

        # Pass the 24 simulated windows to the model at once
        incremental_windows = np.array(incremental_windows)
        batched_probs = model_lstm.predict(incremental_windows, batch_size=SEQUENCE_LEN, verbose=0).flatten()
        fn_max_probs.append(np.max(batched_probs))

    # Classify missed arrests
    for i, idx in enumerate(fn_indices):
        cat = "High-Risk Near Miss (Top 10% of ICU)" if fn_max_probs[i] >= risk_90th else "Sudden Crash"
        fn_records.append({
            'Stay_ID': test_ids[idx],
            'Peak_Risk_Score': round(fn_max_probs[i], 3),
            'Classification': cat
        })

    display(pd.DataFrame(fn_records).sort_values('Peak_Risk_Score', ascending=False).reset_index(drop=True))
```

```python
 # Vectorized Inference for True Positives
tp_indices = np.where((y_test == 1) & (y_test_pred == 1))[0][:4]
if len(tp_indices) > 0:
    sample_seqs = X_test[tp_indices]
    batch_weights = attn_model.predict(sample_seqs, batch_size=64, verbose=0)
    batch_preds = model_lstm.predict(sample_seqs, batch_size=64, verbose=0)

    fig, axes = plt.subplots(len(tp_indices), 1, figsize=(10, 2.5 * len(tp_indices)))
    if len(tp_indices) == 1: axes = [axes] # Handle single plot edge case
    for i, idx in enumerate(tp_indices):
        axes[i].bar(range(SEQUENCE_LEN), batch_weights[i, :, 0], color='teal', alpha=0.8)
        axes[i].set_title(f"Stay {test_ids[idx]} | Arrest Predicted (Prob: {batch_preds[i, 0]:.3f})")
        axes[i].grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

# False Negative Percentile Analysis
fn_indices = np.where((y_test == 1) & (y_test_pred == 0))[0]
if len(fn_indices) > 0:
    # Get 90th percentile risk score of the entire ICU test population
    risk_90th = np.percentile(test_proba, 90)

    fn_records = []
    fn_seqs = X_test[fn_indices]
    fn_max_probs = []

    # Extract maximum probability reached per missed stay
    for seq in fn_seqs:
        probs = [model_lstm.predict(np.vstack([np.zeros((SEQUENCE_LEN-w, seq.shape[1])), seq[SEQUENCE_LEN-w:]])[np.newaxis], verbose=0)[0, 0] for w in range(1, SEQUENCE_LEN+1)]
        fn_max_probs.append(max(probs))

    for i, idx in enumerate(fn_indices):
        cat = "High-Risk Near Miss (Top 10% of ICU)" if fn_max_probs[i] >= risk_90th else "Sudden Crash"
        fn_records.append({'Stay_ID': test_ids[idx], 'Peak_Risk_Score': round(fn_max_probs[i], 3), 'Classification': cat})

    print("\nFalse Negative (Missed Arrest) Analysis:")
    display(pd.DataFrame(fn_records).sort_values('Peak_Risk_Score', ascending=False))
```

```python
# ── SpikeCatcher Layer ─────────────────────────────────
class SpikeCatcher(layers.Layer):
    """Masked causal local-derivative spike detector.
    spike_context: (B, C) - peak spike magnitude per channel over the stay
    spike_score:   (B, T, C) - per-step spike magnitude
    """
    def call(self, inputs, mask=None):
        diff = inputs[:, 1:, :] - inputs[:, :-1, :]
        diff = tf.pad(diff, [[0, 0], [1, 0], [0, 0]])
        spike_score = tf.square(diff)

        if mask is not None:
            mask_f = tf.cast(mask, inputs.dtype)
            spike_score = spike_score * tf.expand_dims(mask_f, -1)

        spike_context = tf.reduce_max(spike_score, axis=1)
        return spike_context, spike_score

    def compute_mask(self, inputs, mask=None):
        return None


# ── Model: LSTM + Attention + SpikeCatcher ─────────────
def build_lstm_spike(seq_len, n_features, hidden=64, dropout=0.3):
    inputs = keras.Input(shape=(seq_len, n_features), name='vital_sequence')

    # FIXED: Wrap tf operations in a Lambda layer to handle KerasTensor
    mask = layers.Lambda(
        lambda x: tf.reduce_any(tf.not_equal(x, 0.0), axis=-1),
        name='mask_generator'
    )(inputs)

    spike_context, spike_score = SpikeCatcher(name='spike_catcher')(inputs, mask=mask)

    x = layers.Masking(mask_value=0.0, name='masking')(inputs)
    x = layers.SpatialDropout1D(dropout, name='spatial_drop_1')(x)
    x = layers.LSTM(hidden, return_sequences=True, name='lstm_1', use_cudnn=False)(x)
    x = layers.LayerNormalization(name='ln_1')(x)
    x = layers.SpatialDropout1D(dropout, name='spatial_drop_2')(x)
    x = layers.LSTM(hidden // 2, return_sequences=True, name='lstm_2', use_cudnn=False)(x)
    x = layers.LayerNormalization(name='ln_2')(x)

    context, attn_weights = MaskedAttention(seed=SEED, name='masked_attention')(x)

    fused = layers.Concatenate(name='attn_plus_spike')([context, spike_context])
    x = layers.Dense(64, activation='relu', name='fc_1')(fused)
    x = layers.Dropout(dropout, seed=SEED)(x)
    out = layers.Dense(1, activation='sigmoid', dtype='float32', name='output')(x)

    model = keras.Model(inputs, out, name='LSTM_Spike')
    spike_model = keras.Model(inputs, [spike_score, attn_weights], name='spike_inspector')
    return model, spike_model


model_lstm, attn_model = build_lstm_spike(SEQUENCE_LEN, len(features), hidden=HIDDEN, dropout=DROPOUT)

model_lstm.compile(
    optimizer=keras.optimizers.Adam(learning_rate=LR),
    loss=keras.losses.BinaryFocalCrossentropy(gamma=2.0, alpha=0.55),
    metrics=[keras.metrics.AUC(name='auc'), keras.metrics.Precision(name='precision'), keras.metrics.Recall(name='recall')]
)

train_dataset = tf.data.Dataset.from_tensor_slices((X_train, y_train)).shuffle(2048, seed=SEED).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
val_dataset   = tf.data.Dataset.from_tensor_slices((X_val, y_val)).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

callbacks = [
    keras.callbacks.EarlyStopping(monitor='val_auc', mode='max', patience=10, restore_best_weights=True),
    keras.callbacks.ReduceLROnPlateau(monitor='val_auc', mode='max', patience=5, factor=0.5, min_lr=1e-5)
]

history = model_lstm.fit(train_dataset, validation_data=val_dataset, epochs=EPOCHS, callbacks=callbacks, verbose=1)


# ── Terminal Spike Timing Analysis (CPR=1 test stays) ──
arrest_idx = np.where(y_test == 1)[0]
X_arrest   = X_test[arrest_idx]
ids_arrest = test_ids[arrest_idx]

spike_scores, attn_weights_arrest = attn_model.predict(X_arrest, batch_size=64, verbose=0)
real_mask = np.any(X_arrest != 0.0, axis=-1)  # (N_arrest, seq_len), True = real (non-pad) step

records = []
for i, sid in enumerate(ids_arrest):
    valid_steps = np.where(real_mask[i])[0]
    per_step_total = spike_scores[i, valid_steps, :].sum(axis=-1)
    peak_t = valid_steps[np.argmax(per_step_total)]
    steps_from_cpr = (SEQUENCE_LEN - 1) - peak_t
    minutes_before_cpr = steps_from_cpr * 30

    records.append({
        'Stay_ID': sid,
        'Minutes_Before_First_CPR': minutes_before_cpr,
        'Peak_Channel': features[np.argmax(spike_scores[i, peak_t, :])],
        'Onset_Type': 'Terminal Spike (≤2hr pre-CPR)'
                        if minutes_before_cpr <= 120 else 'Early/Mid-Window Spike'
    })

spike_df = pd.DataFrame(records)
display(spike_df.sort_values('Minutes_Before_First_CPR'))
```

```python
# ══════════════════════════════════════════════════════
#  ABLATION: Baseline (Attention-only) vs Attention+SpikeCatcher
# ══════════════════════════════════════════════════════

def train_and_eval(model, name, train_ds, val_ds, X_val, y_val, X_test, y_test):
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=LR),
        loss=keras.losses.BinaryFocalCrossentropy(gamma=2.0, alpha=0.55),
        metrics=[keras.metrics.AUC(name='auc'), keras.metrics.Precision(name='precision'), keras.metrics.Recall(name='recall')]
    )
    cbs = [
        keras.callbacks.EarlyStopping(monitor='val_auc', mode='max', patience=10, restore_best_weights=True),
        keras.callbacks.ReduceLROnPlateau(monitor='val_auc', mode='max', patience=5, factor=0.5, min_lr=1e-5)
    ]
    model.fit(train_ds, validation_data=val_ds, epochs=EPOCHS, callbacks=cbs, verbose=0)

    val_proba = model.predict(X_val, verbose=0).flatten()
    best_recall, best_t = 0.0, 0.5
    for th in np.arange(0.01, 1.0, 0.01):
        y_v = (val_proba >= th).astype(int)
        tn, fp, fn, tp = confusion_matrix(y_val, y_v, labels=[0, 1]).ravel()
        if tp == 0: continue
        if (fp / tp) <= MAX_FA_PER_TP and (tp / (tp + fn)) > best_recall:
            best_recall, best_t = tp / (tp + fn), th

    test_proba = model.predict(X_test, verbose=0).flatten()
    y_test_pred = (test_proba >= best_t).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_test, y_test_pred, labels=[0, 1]).ravel()

    return {
        'Model': name,
        'Best_Thresh': round(best_t, 3),
        'Test_AUC': round(roc_auc_score(y_test, test_proba), 3),
        'Test_Recall': round(tp / max(tp + fn, 1), 3),
        'FA_per_TP': round(fp / max(tp, 1), 2),
        'TP': tp, 'FP': fp, 'FN': fn, 'TN': tn
    }

train_dataset = tf.data.Dataset.from_tensor_slices((X_train, y_train)).shuffle(2048, seed=SEED).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
val_dataset   = tf.data.Dataset.from_tensor_slices((X_val, y_val)).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

# Reset seeds before each build so both models get identical init conditions
tf.random.set_seed(SEED)
baseline_model, baseline_attn = build_lstm(SEQUENCE_LEN, len(features), HIDDEN, DROPOUT)
baseline_result = train_and_eval(baseline_model, 'Baseline (Attention only)', train_dataset, val_dataset, X_val, y_val, X_test, y_test)

tf.random.set_seed(SEED)
spike_model_full, spike_attn = build_lstm_spike(SEQUENCE_LEN, len(features), HIDDEN, DROPOUT)
spike_result = train_and_eval(spike_model_full, 'Attention + SpikeCatcher', train_dataset, val_dataset, X_val, y_val, X_test, y_test)

ablation_df = pd.DataFrame([baseline_result, spike_result])
print("=" * 70)
print(" ABLATION: SpikeCatcher Contribution")
print("=" * 70)
display(ablation_df)
```
